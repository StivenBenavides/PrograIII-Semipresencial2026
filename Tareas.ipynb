```python

# ============================================================
#              PROGRAMA COMPLETO DE CONVERSORES
# ============================================================

ejecutando = True


# ============================================================
#              FUNCIÓN GENERAL DE CONVERSIÓN
# ============================================================

def conversor(nombre, unidades, factores):

    while True:

        print("\n========================================")
        print(f"        CONVERSOR DE {nombre.upper()}")
        print("========================================")

        for numero, unidad in unidades.items():
            print(f"{numero}. {unidad}")

        # ----------------------------------------------------
        # Elegir unidad de origen
        # ----------------------------------------------------

        while True:
            try:
                origen = int(input("\n¿De qué unidad desea convertir?: "))

                if origen in unidades:
                    break
                else:
                    print("Opción no válida. Intente nuevamente.")

            except ValueError:
                print("Debe ingresar un número.")

        # ----------------------------------------------------
        # Elegir unidad de destino
        # ----------------------------------------------------

        while True:
            try:
                destino = int(input("¿A qué unidad desea convertir?: "))

                if destino in unidades:
                    break
                else:
                    print("Opción no válida. Intente nuevamente.")

            except ValueError:
                print("Debe ingresar un número.")

        # ----------------------------------------------------
        # Ingresar cantidad
        # ----------------------------------------------------

        while True:
            try:
                valor = float(
                    input("Ingrese el valor que desea convertir: ")
                )

                if valor < 0:
                    print("El valor no puede ser negativo.")
                else:
                    break

            except ValueError:
                print("Debe ingresar un número válido.")

        # ====================================================
        # CONVERSIÓN DE MONEDAS
        # ====================================================

        if nombre == "Monedas":

            resultado = (
                valor * factores[destino]
            ) / factores[origen]

        # ====================================================
        # CONVERSIÓN DE LAS DEMÁS UNIDADES
        # ====================================================

        else:

            unidad_base = valor * factores[origen]

            resultado = unidad_base / factores[destino]

        # ----------------------------------------------------
        # Mostrar resultado
        # ----------------------------------------------------

        print("\n========================================")
        print("        RESULTADO DE LA CONVERSIÓN")
        print("========================================")

        print(
            f"{valor} {unidades[origen]} = "
            f"{resultado:.6f} {unidades[destino]}"
        )

        # ====================================================
        # OPCIONES DESPUÉS DE LA CONVERSIÓN
        # ====================================================

        while True:

            print("\n11. Realizar otra conversión")
            print("12. Salir")
            print("13. Volver al menú principal")

            try:
                continuar = int(
                    input("\nSeleccione una opción: ")
                )

                if continuar == 11:

                    break

                elif continuar == 12:

                    print("\nPrograma finalizado.")
                    return "salir"

                elif continuar == 13:

                    print("\nVolviendo al menú principal...")
                    return "menu"

                else:

                    print(
                        "Opción no válida. "
                        "Intente nuevamente."
                    )

            except ValueError:

                print("Debe ingresar un número.")


# ============================================================
#                    CONVERSOR DE MONEDAS
# ============================================================

monedas = {
    1: "Dólares",
    2: "Euros",
    3: "Libras esterlinas",
    4: "Yenes",
    5: "Dólares canadienses",
    6: "Pesos mexicanos",
    7: "Quetzales",
    8: "Lempiras",
    9: "Colones costarricenses",
    10: "Córdobas"
}


# ============================================================
# TASAS DE MONEDAS
#
# MONEDA DE REFERENCIA: EURO
#
# Los valores representan:
#
# 1 EUR = cantidad de moneda indicada
# ============================================================

factores_monedas = {

    1: 1.1605,
    2: 1.0000,
    3: 0.85608,
    4: 184.62,
    5: 1.6098,
    6: 19.8072,
    7: 8.7516,
    8: 30.7108,
    9: 515.017,
    10: 42.53
}


# ============================================================
#                    CONVERSOR DE LONGITUD
# ============================================================

longitud = {
    1: "Kilómetros",
    2: "Metros",
    3: "Centímetros",
    4: "Milímetros",
    5: "Millas",
    6: "Yardas",
    7: "Pies",
    8: "Pulgadas",
    9: "Millas náuticas",
    10: "Micrómetros"
}

factores_longitud = {
    1: 1000,
    2: 1,
    3: 0.01,
    4: 0.001,
    5: 1609.344,
    6: 0.9144,
    7: 0.3048,
    8: 0.0254,
    9: 1852,
    10: 0.000001
}


# ============================================================
#                      CONVERSOR DE MASA
# ============================================================

masa = {
    1: "Toneladas",
    2: "Kilogramos",
    3: "Hectogramos",
    4: "Decagramos",
    5: "Gramos",
    6: "Decigramos",
    7: "Centigramos",
    8: "Miligramos",
    9: "Libras",
    10: "Onzas"
}

factores_masa = {
    1: 1000000,
    2: 1000,
    3: 100,
    4: 10,
    5: 1,
    6: 0.1,
    7: 0.01,
    8: 0.001,
    9: 453.59237,
    10: 28.349523125
}


# ============================================================
#                 CONVERSOR DE ALMACENAMIENTO
# ============================================================

almacenamiento = {
    1: "Bytes",
    2: "Kilobytes",
    3: "Megabytes",
    4: "Gigabytes",
    5: "Terabytes",
    6: "Petabytes",
    7: "Exabytes",
    8: "Bits",
    9: "Kilobits",
    10: "Megabits"
}

factores_almacenamiento = {
    1: 1,
    2: 1024,
    3: 1024**2,
    4: 1024**3,
    5: 1024**4,
    6: 1024**5,
    7: 1024**6,
    8: 0.125,
    9: 1024 / 8,
    10: (1024**2) / 8
}


# ============================================================
#                    CONVERSOR DE TIEMPO
# ============================================================

tiempo = {
    1: "Microsegundos",
    2: "Milisegundos",
    3: "Segundos",
    4: "Minutos",
    5: "Horas",
    6: "Días",
    7: "Semanas",
    8: "Meses",
    9: "Años",
    10: "Décadas",
    11: "Siglos"
}

factores_tiempo = {
    1: 0.000001,
    2: 0.001,
    3: 1,
    4: 60,
    5: 3600,
    6: 86400,
    7: 604800,
    8: 2592000,
    9: 31536000,
    10: 315360000,
    11: 3153600000
}


# ============================================================
#                    CONVERSOR DE VOLUMEN
# ============================================================

volumen = {
    1: "Litros",
    2: "Mililitros",
    3: "Metros cúbicos",
    4: "Centímetros cúbicos",
    5: "Galones",
    6: "Pintas",
    7: "Cuartos",
    8: "Tazas",
    9: "Onzas líquidas",
    10: "Decilitros"
}

factores_volumen = {
    1: 1,
    2: 0.001,
    3: 1000,
    4: 0.001,
    5: 3.785411784,
    6: 0.473176473,
    7: 0.946352946,
    8: 0.2365882365,
    9: 0.0295735296,
    10: 0.1
}


# ============================================================
#                      CONVERSOR DE ÁREA
# ============================================================

area = {
    1: "Kilómetros cuadrados",
    2: "Hectáreas",
    3: "Metros cuadrados",
    4: "Decímetros cuadrados",
    5: "Centímetros cuadrados",
    6: "Milímetros cuadrados",
    7: "Millas cuadradas",
    8: "Yardas cuadradas",
    9: "Pies cuadrados",
    10: "Acres"
}

factores_area = {
    1: 1000000,
    2: 10000,
    3: 1,
    4: 0.01,
    5: 0.0001,
    6: 0.000001,
    7: 2589988.110336,
    8: 0.83612736,
    9: 0.09290304,
    10: 4046.8564224
}


# ============================================================
#                     MENÚ PRINCIPAL
# ============================================================

while ejecutando:

    print("\n")
    print("========================================")
    print("           MENÚ PRINCIPAL")
    print("========================================")
    print("1. Monedas")
    print("2. Longitud")
    print("3. Masa")
    print("4. Almacenamiento")
    print("5. Tiempo")
    print("6. Volumen")
    print("7. Área")
    print("8. Salir")
    print("========================================")

    try:

        opcion_principal = int(
            input(
                "\nSeleccione el conversor "
                "que desea utilizar: "
            )
        )

    except ValueError:

        print("\nDebe ingresar un número.")
        continue


    # ========================================================
    # MONEDAS
    # ========================================================

    if opcion_principal == 1:

        resultado = conversor(
            "Monedas",
            monedas,
            factores_monedas
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # LONGITUD
    # ========================================================

    elif opcion_principal == 2:

        resultado = conversor(
            "Longitud",
            longitud,
            factores_longitud
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # MASA
    # ========================================================

    elif opcion_principal == 3:

        resultado = conversor(
            "Masa",
            masa,
            factores_masa
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # ALMACENAMIENTO
    # ========================================================

    elif opcion_principal == 4:

        resultado = conversor(
            "Almacenamiento",
            almacenamiento,
            factores_almacenamiento
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # TIEMPO
    # ========================================================

    elif opcion_principal == 5:

        resultado = conversor(
            "Tiempo",
            tiempo,
            factores_tiempo
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # VOLUMEN
    # ========================================================

    elif opcion_principal == 6:

        resultado = conversor(
            "Volumen",
            volumen,
            factores_volumen
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # ÁREA
    # ========================================================

    elif opcion_principal == 7:

        resultado = conversor(
            "Área",
            area,
            factores_area
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # SALIR
    # ========================================================

    elif opcion_principal == 8:

        print("\nPrograma finalizado.")
        ejecutando = False


    # ========================================================
    # OPCIÓN NO VÁLIDA
    # ========================================================

    else:

        print("\nOpción no válida. Intente nuevamente.")


# ============================================================
#                       FIN DEL PROGRAMA
# ============================================================

print("\nGracias por utilizar el programa.")
```
